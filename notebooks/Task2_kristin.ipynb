# Task 2 - Reasoning through time (and the frame problem)
In your second task, the world changes over time and is explicitly represented. The WM now represents the state of a world at a moment in time and actions change this world. Use a simple pick-and-place world: A hand and some objects at locations. The hand can move, grab, and release. 

An action is a production whose firing changes the WM. It makes some facts true (an add set) and some false (a delete set) and changes the time of the world. 

Every fact carries a time step as its last argument and is treated as a belief that holds at that time. E.g., (at, obj1, A, 0) translates as “obj1 is at A at step 0”. A fact holds at a step and stays true into the next step onl because it is carried there, which mean that in pcinciple any fact can be changed by an action. 

If an action fires at time ‘t’ it means that its conditions are read at ‘t’ and its effects land at ‘t+1’. It names what it makes true (‘add’) and what it end (‘delete’). 

Example:
action grab(?o, ?loc) at step ?t:
conditions: at(hand, ?loc, ?t), at(?o, ?loc, ?t), empty(hand, ?t)
Add at ?t+1: holding(?o)
delete: empty(hand), at(?o, ?loc)

## 2.1 Carrying knowledge through time
Design the representation and WM management that produce the next state from the current one when an action fires. Remember, the interesting part is not only the add/delete functionality, but rather everything the action does not mention. If the hand grabs object 1, object 2 must still be where it was, the other locations must still exist, etc.

Make this concrete in two steps and record both:

### 2.1.1 Record what happens
First, produce the next state by recording only what the action makes true at ‘t+1’. Observe what happens to the rest of the world at ‘t+1’.

### 2.1.2 Persistence assumption
Next, introduce a persistence assumption: A fact that is true at step ‘t’ is re-asserted at ‘t+1’ unless an action deletes it. Implement this as a generic “advnace one step” in the engine that you have written. There should be no reference to any particular predicate.

This persistence assumption is your answer to the frame problem. In the report, explain why the naive alternative (stating for every fact and every action that the fact still exists afterward) does not scale.
