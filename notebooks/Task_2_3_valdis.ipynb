```python
def at_time(facts, t):
    return {fact for fact in facts if fact[-1] == t}

def contents(facts, t):
    """Untimed view for comparing two states without their timestamps."""
    return {fact[:-1] for fact in at_time(facts, t)}

def show_state(facts, t):
    state = at_time(facts, t)
    print(f"\nSTATE t={t} ({len(state)} facts)")
    for fact in sorted(state, key=repr):
        print(" ", fact)
```

```python
import importlib
sherlock = importlib.import_module("reasoning.engine")
from reasoning.domains import ACTIONS, TEST_2_FACTS, SCHEMAS
```

```python
WorkingMemory = sherlock.WorkingMemory
```

```python
wm = WorkingMemory(TEST_2_FACTS)
show_state(wm.facts, 0)
```

```python
# I want the hand to go to B
# Input is then the goal
goal = [("at", "hand", "B")]

actions = sherlock.plan(wm, SCHEMAS, goal, t=0)

assert actions == [('move', 'A', 'B')]
print(f"goal {goal} achieved")
```

```python
# I want the hand to go to C
# Input is then the goal
goal = [("at", "hand", "C")]

actions = list(sherlock.plan(wm, SCHEMAS, goal, t=0))

assert actions == [('move', 'A', 'B'), ('move', 'B', 'C')]
print(f"goal {goal} achieved")
```
