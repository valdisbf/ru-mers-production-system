# The evidence courier — Jupyter display example

The module renders timed working-memory facts as a fixed floor plan. These first pictures are **hand-built display fixtures**, not an action engine or proof that the assignment works. Connect your own engine in the final section.

Keep `world_view.py` beside this notebook for this standalone example.

```python
from world_view import WorldView

TASK2_INITIAL_FACTS = {('at', 'letter', 'library', 0), ('location', 'garden', 0), ('location', 'terrace', 0), ('object', 'letter_opener', 0), ('at', 'letter_opener', 'study', 0), ('passage', 'hallway', 'gallery', 0), ('passage', 'foyer', 'garden', 0), ('object', 'sample_vial', 0), ('location', 'foyer', 0), ('passage', 'library', 'hallway', 0), ('object', 'key', 0), ('passage', 'hallway', 'library', 0), ('at', 'sample_vial', 'laboratory', 0), ('object', 'letter', 0), ('passage', 'gallery', 'laboratory', 0), ('empty', 'sherlock', 0), ('evidence', 'letter_opener', 0), ('at', 'key', 'terrace', 0), ('location', 'hallway', 0), ('passage', 'hallway', 'foyer', 0), ('location', 'study', 0), ('passage', 'gallery', 'study', 0), ('evidence', 'sample_vial', 0), ('passage', 'study', 'gallery', 0), ('evidence', 'letter', 0), ('location', 'gallery', 0), ('passage', 'laboratory', 'gallery', 0), ('passage', 'garden', 'terrace', 0), ('location', 'library', 0), ('at', 'sherlock', 'foyer', 0), ('passage', 'foyer', 'hallway', 0), ('location', 'laboratory', 0), ('passage', 'gallery', 'hallway', 0), ('passage', 'garden', 'foyer', 0), ('passage', 'terrace', 'garden', 0)}

view = WorldView(TASK2_INITIAL_FACTS)
view.show(TASK2_INITIAL_FACTS, 0)
```

## A carrying state

A display fixture for Sherlock at the study holding the letter opener. Other objects, room declarations and passages still exist. The letter opener has no separate `at` fact while held.

```python
# Illustrative state only; these updates stand in for your engine output.
carried = {f[:-1] + (1,) for f in TASK2_INITIAL_FACTS
           if f not in {("at", "sherlock", "foyer", 0),
                        ("empty", "sherlock", 0),
                        ("at", "letter_opener", "study", 0)}}
carried |= {("at", "sherlock", "study", 1), ("holding", "letter_opener", 1)}
history = TASK2_INITIAL_FACTS | carried
view.show(history, 1, title="Example display: evidence collected")
```

## A delivered state

Another independent fixture: all evidence is in the foyer. The terrace key is still there. These fixtures skip intermediate moves; your real demo must run its actual actions.

```python
delivered = {f[:-1] + (2,) for f in TASK2_INITIAL_FACTS
             if not (f[0] == "at" and f[1] in {"letter_opener", "letter", "sample_vial"})}
delivered |= {("at", obj, "foyer", 2) for obj in ("letter_opener", "letter", "sample_vial")}
view.show(TASK2_INITIAL_FACTS | delivered, 2, title="Example display: delivery complete")
```

## Effects only: the missing world

Only the new holding fact reaches the next state. The static map remains as reference context, with dim room cards and dashed links, but no historical positions are reused. The robot location and resting object locations are unknown.

```python
effects_only = TASK2_INITIAL_FACTS | {("holding", "letter_opener", 1)}
view.show(effects_only, 1, title="Effects only: where did the world go?")
```
